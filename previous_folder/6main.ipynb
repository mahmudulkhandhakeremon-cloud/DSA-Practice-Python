Insertion Sort

Insertion Sort is a simple sorting algorithm that takes one element at a time and inserts it into its correct position in the sorted part of the array.

Easy to remember:
Take → Compare → Shift → Insert


```python
my_array = [5, 2, 4]

n = len(my_array)

for i in range(1, n):
    insert_index = i
    current_value = my_array.pop(i)

    for j in range(i - 1, -1, -1):
        if my_array[j] > current_value:
            insert_index = j

    my_array.insert(insert_index, current_value)

print("Sorted array:", my_array)
```

Insertion Sort  Improvement Solution

```python
my_array = [5, 2, 4]

n = len(my_array)

for i in range(1, n):
    insert_index = i
    current_value = my_array[i]

    for j in range(i - 1, -1, -1):
        if my_array[j] > current_value:
            my_array[j + 1] = my_array[j]
            insert_index = j
        else:
            break

    my_array[insert_index] = current_value

print("Sorted array:", my_array)
```

N.B: Take a value → shift the larger values to the right → insert the value into the empty spot.    

The main operation in this code is:

if my_array[j] > current_value:
    my_array[j + 1] = my_array[j]


In other words, shift the larger value one position to the right.

Time Complexity: O(n²)
