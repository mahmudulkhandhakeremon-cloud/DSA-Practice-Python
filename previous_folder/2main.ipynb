To calculate the Fibonacci Number Algorithm by using A For Loop

```python
num1=0
num2=1

print(num1)
print(num2)

for fibonacci in range(18): ##  range(n)-->n is any type of numbers
    newfibonacci=num1+num2
    print(newfibonacci)
    num2=num1
    num1=newfibonacci
    
```

Implementing Recursion to calculate fibonacci numbers

```python
print(0)
print(1)

count=2

def fibonacci(num1,num2):
    global count
    if count <=19:
        newfibonacci= num1+num2
        print(newfibonacci)
        num2=num1
        num1=newfibonacci
        count +=1
        fibonacci(num1,num2)
    else:
        return
fibonacci(1,0)
```

Finding The n(anonymous)fibonacci number Using recursion

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```
