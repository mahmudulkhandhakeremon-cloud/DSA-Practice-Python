To calculate the minimum values of the Array

```python
my_array=[7,12,5,7,8,9,2]
minval=my_array[0]
for i in my_array:
    if i <minval:
       minval=i
print("Lowest Value:",minval)
```

To Calculate the Maximum values of the Array

```python
my_array=[7,12,5,7,8,9,2]
maxval=my_array[0]
for i in my_array:
    if i >maxval:
       maxval=i
print("highest Value:",maxval)
```
